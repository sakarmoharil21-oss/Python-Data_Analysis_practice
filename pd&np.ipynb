```python

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Dept": ["IT", "IT", "HR", "HR", "IT", "HR", "IT", "HR"],
    "Marks": [85, 72, 91, 65, 78, 88, 95, 74],
    "attendence": [92, 76, 95, 68, 82, 90, 98, 72],
    "projects": [4, 2, 5, 1, 3, 4, 6, 2]
})

# Calculate weighted score
df["score"] = (
    df["Marks"] * 0.5
    + df["attendence"] * 0.3
    + df["projects"] / df["projects"].max() * 100 * 0.2
)

# Department-wise rank
df["rank"] = df.groupby("Dept")["score"].rank(ascending=False)

# Department-wise best score
df["dept_best"] = df.groupby("Dept")["score"].transform("max")

# Department-wise average score
df["dept_avg"] = df.groupby("Dept")["score"].transform("mean")

# Compare each employee's score with department average
df["Performance"] = np.where(
    df["score"] >= df["dept_avg"],
    "Above average",
    "Below average"
)

# Employees above department average with at least 3 projects
# Select the highest scorer from each department
top = (
    df[
        (df["score"] > df["dept_avg"])
        & (df["projects"] >= 3)
    ]
    .groupby("Dept")["score"]
    .idxmax()
)

print("Top qualifying employee from each department:")
print(df.loc[top])

# Performance proportions within each department
performance = df.groupby("Dept")["Performance"].value_counts(
    normalize=True
)
print("\nDepartment-wise performance proportions:")
print(performance)

# Most common performance category in each department
result = performance.groupby(level=0).idxmax()
print("\nMost common performance category:")
print(result)

# Department summary: average score, attendance, total projects
op = df.groupby("Dept").agg({
    "score": "mean",
    "attendence": "mean",
    "projects": "sum"
})
print("\nDepartment summary:")
print(op)

# Departments with average score and attendance above 75
r = df.groupby("Dept").agg({
    "score": "mean",
    "attendence": "mean"
})

rr = (r["score"] > 75) & (r["attendence"] > 75)
print("\nDepartments meeting both conditions:")
print(rr)
print(r.loc[rr])

# Multiple aggregations create MultiIndex columns
r = df.groupby("Dept").agg({
    "score": ["mean", "max"],
    "attendence": "mean",
    "projects": "sum"
})

# Sort by average score, then maximum score, descending
rr = r.sort_values(
    by=[("score", "mean"), ("score", "max")],
    ascending=False
)
print("\nFinal sorted department summary:")
print(rr)

```

```python

import pandas as pd
import numpy as np

data = {
    "Name": ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Sara", "Vihaan", "Anaya", "Arjun", "Isha"],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai", "Pune", "Nashik", "Mumbai", "Pune", "Nashik"],
    "Age": [21, 22, 20, 23, 21, 24, 22, 20, 25, 23],
    "Salary": [45000, 52000, 38000, 60000, 48000, 70000, 55000, 42000, 75000, 58000],
    "Experience": [1, 2, 0, 3, 1, 4, 2, 0, 5, 3],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR", "IT", "Finance"]
}

df = pd.DataFrame(data)

# 1. Average salary by department
r = df.groupby("Department")["Salary"].mean()
rr = r[r > 50000]
print(rr)

# 2. Department with the highest average salary
top = df.groupby("Department")["Salary"].mean().idxmax()
print(top)

# 3. Difference between employee salary and department average
df["s_vs_avg"] = (
    df["Salary"]
    - df.groupby("Department")["Salary"].transform("mean")
)
print(df)

# 4. Rank salaries within each department
df["SalaryR"] = df.groupby("Department")["Salary"].rank(
    ascending=False
)

# 5. Highest-paid employee in each department
r = df.groupby("Department")["Salary"].idxmax()
print(df.loc[r])

# 6. Department summary
s = df.groupby("Department").agg({
    "Salary": ["mean", "max"],
    "Experience": "mean"
})

print(s.sort_values(("Salary", "mean"), ascending=True))

```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aarav", "Meera", "Kabir", "Diya", "Arjun", "Anaya", "Vihaan", "Ishita"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "Salary": [55000, 48000, 62000, 58000, 51000, 67000, 60000, 53000],
    "Attendance": [92, 85, 78, 95, 88, 91, 82, 89],
    "Projects": [4, 3, 5, 4, 2, 6, 5, 3],
    "Experience": [2, 1, 3, 4, 2, 5, 3, 2]
}

df = pd.DataFrame(data)
ns=(df["Salary"]-df["Salary"].min())/(df["Salary"].max()-df["Salary"].min())
df["score"]=ns*0.40+df["Attendance"]*0.30 + df["Projects"]*0.30
df["performance_level"]=np.where(df["score"]>=30,"Excellent",np.where(df["score"]>=27,"good","needs improvement"))

print(df.groupby("Department")["score"].mean())
top=df.groupby("Department")["score"].idxmax()
print(df.loc[top,"Name"])
df["salary vs deptavg"]=df["Salary"]-df.groupby("Department")["Salary"].transform("mean")
print(df)
df["above"]=np.where(df["Salary"]>df.groupby("Department")["Salary"].transform("mean"),True,False)
print(df)
topp=df["score"].idxmax()
print(df.loc[topp,["Name","Department"]])
df["dept perf avg"]=df.groupby("Department")["score"].transform("mean")
df["Performance diff"]=df["score"]-df["dept perf avg"]
df["performance"]=np.select([df["Performance diff"]>0,df["Performance diff"]<0],["Above avg","below avg"],default ="avg")
print(df)
acc=df.groupby("Department")["Salary"].idxmax()
print(df.loc[acc,["Name","Department","Salary"]])
ad=df.groupby("Department")["Salary"].mean()
tops=ad.idxmax()
print(ad.loc[tops])
acs=df.groupby("Department")["above"].mean()*100
print(acs)
```
