```python

import pandas as pd

data = {
    "Name": ["A", "B", "C", "D", "E"],
    "Age": [20, 21, 19, 22, 25],
    "Marks": [85, 92, 98, 90, 77]
}

df = pd.DataFrame(data)


print("Name and Age:")
print(df[["Name", "Age"]])

print("\nAge greater than 20 and less than 22:")
print(df[(df["Age"] > 20) & (df["Age"] < 22)])
print("\nAge greater than 20 or Marks greater than 90:")
print(df[(df["Age"] > 20) | (df["Marks"] > 90)])

df["bonus"] = df["Age"] * 10

print("\nAfter adding bonus:")
print(df)


df.insert(1, "id", [1, 2, 3, 4, 5])

print("\nAfter inserting ID:")
print(df)


df.drop(columns=["Marks"], inplace=True)

print("\nAfter dropping Marks:")
print(df)

```

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Harsh","Rahul","sneha","priya","Raj"],
    "Age":[20,22,19,23,24,21],
    "Marks":[84,75,66,98,87,54]
})
df["D_Marks"]=df["Marks"]*2
print(df)
df.drop("Age",axis=1,inplace=True)
print(df)
df.drop(3,axis=0,inplace=True)
print(df)
```

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Harsh","Rahul","None","priya","Raj"],
    "Age":[20,None,19,23,24,21],
    "Marks":[84,75,None,98,87,54]
})
df.sort_values("Age",ascending=False,inplace=True)
print(df)
df.sort_values("Marks",inplace=True)
print(df)
df.sort_values(["Age","Marks"],ascending=[True,False],inplace=True)
print(df)
print(df["Name"].nunique())
print(df.isna().sum())
df["Marks"]=df["Marks"].interpolate(method="linear")
print(df)
```

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["A","B","C","D","E","F"],
    "Department":["IT","HR","IT","HR","SALES","SALES"],
    "SALARY":[50000,40000,60000,45000,55000,65000]
})
print(df[(df["SALARY"]>50000) & (df["Department"]=="IT")])
```

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Harsh","Rahul","None","priya","Raj"],
    "Dept":["IT","HR","IT","HR","Sales","Sales"],
    "Salary":[50000,40000,60000,45000,55000,65000]
})
print(df.groupby("Dept")["Salary"].mean())
print(df.groupby("Dept")["Salary"].sum())
print(df.groupby("Dept")["Salary"].max())
print(df.groupby("Dept")["Salary"].min())

print(df.groupby("Dept")["Salary"].agg(["mean","max","min","count"]))
```
