```python
import numpy as np
arr=np.array([[12,5,8],
              [20,7,15],
              [3,18,10]])
mask=arr>10
print(arr[mask])
mask=arr%2==0
print(arr[mask])
mask=arr%5==0
print(arr[mask])
arr[arr%2==0]=-1
arr=np.array([[12,5,8],
              [20,7,15],
              [3,18,10]])
arr[arr>10]
print(np.count_nonzero(arr>10))
arr=np.array([[12,5,8],
              [20,7,15],
              [3,18,10]])

arr[arr>10]
print(np.argwhere(arr>10))
```

```python
import numpy as np
arr=np.array([10,20,30,40,50,60])
print(arr[[1,3,5]])
ar=np.array([[10,20,30],
              [40,50,60],
              [70,80,90]])
print(ar[[0,2]])
print(ar[[0,1,2],[1,2,0]])
print(ar[[0,1,2],[2,1,0]])
abc=np.array([[5,10,15],
              [20,25,30],
              [35,40,45]])
print(abc[[0,1,2],[0,1,2]])
print(abc[[0,1,2],[2,0,1]])
```

```python
import numpy as np
arr=np.array([[12,5,8,25],
              [3,18,7,40],
              [15,2,30,10]])
print(np.where(arr>10))
print(np.where(arr%2==0))
ar=np.where(arr<10,0,arr)
print(ar)
abc=np.where(arr%2==0, 1 ,0)
print(abc)
new=np.where(arr>=10,arr,-1)
print(new)
```

```python
import numpy as np

a=np.array([[1,2],
            [3,4]])
b=np.array([[5,6],
            [7,8]])
print(np.concatenate((a,b)))
print(np.concatenate((a,b),axis=1))
print(np.vstack((a,b)))
print(np.hstack((a,b)))

arr=np.array([8,2,5,2,1,8,4])
print(np.sort(arr))
print(np.unique(arr))
print(np.argsort(arr))
```
